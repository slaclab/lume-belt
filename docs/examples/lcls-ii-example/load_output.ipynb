```python
import numpy as np
import matplotlib.pyplot as plt
from pmd_beamphysics.units import mec2
```

```python
label = 108
data = np.loadtxt("./fort." + str(label + 1))
data_c = np.loadtxt("./fort." + str(label))
E0 = 100e6
```

```python
zc = data_c[:, 0]
current = data_c[:, 2]
```

```python
z = data[:, 0]
dgamma = data[:, 1]
weight = data[:, 2]
dE_E0 = data[:, 3]
gamma0 = dgamma / dE_E0
gamma = gamma0 + dgamma
```

```python
np.mean(gamma0 * mec2)
```

```python
print("Mean Energy = ", np.mean(gamma * mec2 / 1e9), "GeV")
print("Mean z = ", np.mean(z) * 1e3, "mm")
print("sigma_z = ", np.std(z) * 1e3, "mm")
print("sigma_E/<E> = ", np.std(gamma) / np.mean(gamma0) * 1e2, "%")
print("Peak current = ", np.max(current) / 1e3, "kA")
```

```python
plt.plot(z * 1e3, gamma * mec2 / 1e6, ".")
plt.xlabel("z (mm)")
plt.ylabel("E(MeV)")
```
